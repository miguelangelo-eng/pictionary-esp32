```python
import os
import urllib.request
import urllib.parse

CATEGORIAS = [
    "apple", "bicycle", "car", "cat", "clock",
    "eye", "fish", "house", "key", "light bulb",
    "rainbow", "star", "sun", "tree", "umbrella",
] 
URL_BASE = "https://storage.googleapis.com/quickdraw_dataset/full/numpy_bitmap/"
PASTA = "data"

os.makedirs(PASTA, exist_ok=True)

def baixar(categoria):
    destino = os.path.join(PASTA, f"{categoria}.npy")

    # TODO 1: se o arquivo 'destino' já existe, imprima uma mensagem e saia da função
    #         dica: os.path.exists(...)  e  return
    if os.path.exists(destino):
        print(f"endereço {destino} já existe.")

    # TODO 2: monte a URL completa (URL_BASE + nome da categoria convertido com quote)
    url = URL_BASE + urllib.parse.quote(categoria) + ".npy"

    print(f"baixando {categoria}...")
    # TODO 3: baixe a URL para 'destino'
    #         dica: urllib.request.urlretrieve(url, destino)
    urllib.request.urlretrieve(url, destino)

    tamanho_mb = os.path.getsize(destino) / 1e6
    print(f"  ok: {tamanho_mb:.0f} MB")

# teste com UMA categoria primeiro:
baixar("cat")

# TODO 4: use um loop para baixar todas as categorias
for categoria in CATEGORIAS:
    baixar(categoria)
```
